```python
def map_tick(x):
    if x % 6 == 1:
        return x // 6 + 1
    elif x % 6 == 4:
        return x // 6 + 2
    elif x % 6 == 2:
        return x // 6 + 3
    elif x % 6 == 5:
        return x // 6 + 4
    elif x % 6 == 3:
        return x // 6 + 5
    elif x % 6 == 0:
        return x // 6 + 6
```

```python
from importlib.resources import path
import mpmath as mpf
epsilon = 2
dG_1 = 2
dG_2 = 2
k_bT = 1

epsilon_1 = osc_midpoint_six_states(alpha_32, alpha_21, alpha_65, alpha_54, 0, 0, dG_1, dG_2, gamma, 0,0,k_bT)  + 3
epsilon_2 = osc_midpoint_six_states(alpha_32, alpha_21, alpha_65, alpha_54, 0, 0, dG_1, dG_2, gamma, 0,0,k_bT) - 3
alpha_21 = 1
alpha_12 = alpha_21 * mpf.exp((dG_1 + epsilon_1 )/ k_bT)
alpha_32 = 1
alpha_23 = alpha_32 * mpf.exp((dG_2 - epsilon_1 )/ k_bT)

alpha_54 = 1
alpha_45 = alpha_54 * mpf.exp((dG_2 + epsilon_2 )/ k_bT)
alpha_65 = 1
alpha_56 = alpha_65 * mpf.exp((dG_1 - epsilon_2 )/ k_bT)

alpha_14 = 0.1
alpha_41 = 0.1
alpha_25 = alpha_14
alpha_52 = alpha_41
alpha_36 = alpha_14
alpha_63 = alpha_41

gamma_1 = 10000
gamma_2 = 10000





P = np.array([[0, alpha_21, gamma_1, alpha_41, 0 , 0],
                [alpha_12, 0, alpha_32, 0, alpha_52, 0],
                [0, alpha_23, 0, 0, 0, alpha_63],
                [alpha_14, 0, 0, 0, alpha_54, gamma_2],
                [0, alpha_25, 0, alpha_45, 0, alpha_65],
                [0, 0, alpha_36, 0, alpha_56, 0]], dtype=float)  

i = 0;
path_trajectories_slow = []
path_times_slow = []
for _ in range(10):
    path_trajectory = [1]
    path_time = [0]
    t = 0
    state = 0
    cycles = 0
    while t < 20:
        rates = P[:, state]
        k_tot = np.sum(rates)
        rho_1, rho_2 = np.random.rand(2)
        state_candidate = np.random.choice(len(rates))
        if rates[state_candidate] >= rho_1 * k_tot:
            path_trajectory.append(6 * cycles + map_tick(state+1))   
            if state_candidate == 0 and state == 2:
                cycles += 1
            if state_candidate == 3 and state == 5:
                cycles += 1
            state = state_candidate
            path_trajectory.append(6 * cycles + map_tick(state+1))
            t -= np.log(rho_2) / k_tot
            path_time.append(t)
            path_time.append(t)
    path_trajectories_slow.append(path_trajectory)
    path_times_slow.append(path_time)
```
